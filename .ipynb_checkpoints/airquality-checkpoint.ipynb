```python
#cleaning data

df.isnull() #both same

df.isnull().sum()



df.dropna()

df.dropna(axis = 1)

df.fillna("unknown")

# cleaned data storing

cleaned_data = df.copy()
cleaned_data['age'] = cleaned_data["age"].fillna(cleaned_data['age'].mean())

df.ffill()

df.bfill()
```

```python

df2 = df.copy()

df2["Tax"] = df2["income"].apply(lambda x : "20%" if x >= 50000 else "10%")

gender_map = {"Male" : 'M', "Female": 'F', "Unknown": 'U'}
df2["gender"] = df2["gender"].map(gender_map)

df2 = df2.assign(new_income = df2["income"]*1.1)

df2["country"] = df2["country"].replace("USA", "US")

df2.columns = ['id','name','age','country','gender','income','Tax', 'new_income']
df2.rename(columns= {"income":"Salary"})
df2.rename(index={4:'Fourth'})

sorted_data = df2.sort_values(["income", "age"], ascending=False)
sorted_data.sort_index()

sorted_data.reset_index()
sorted_data.reset_index(drop=True)

sorted_data["Ranking"] = sorted_data["income"].rank(ascending=False, method ='max')

sorted_data = sorted_data[['id','name','age','country','gender','income', 'new_income','Tax']]

sorted_data
```
