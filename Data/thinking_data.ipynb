```python
import json
```

```python
#loading the data
def load_data(filename):
    with open(filename, "r") as f:
        data = json.load(f)

    return data
```

```python
data = load_data("Data/data.json")
print(data)
print(type(data))
```

```python
#cleaning the data

def clean_data(data):
    cleaned_data = []
    unique_users = set()
    text_to_num = {"one": 1, "two": 2, "three": 3, "four": 4, "five": 5}
    
    for user in data:
        #clean rating
        raw_rating = str(user["rating"]).strip().lower()
        if (raw_rating in text_to_num):
            raw_rating = text_to_num[raw_rating]
        user["rating"] = raw_rating

        #Handling missing values
        raw_age = user.get("age")
        if(raw_age == None):
            user["age"] = None
            
        if(user["name"].strip() in unique_users):
            continue
            
        unique_users.add(user["name"].strip())
        cleaned_data.append(user)
    return cleaned_data
```

```python
cleaned_data = clean_data(data)
for user in cleaned_data:
    print(user)
```

```python
#get meaningful insights from data
def get_insights(cleaned_data):
    
    total_rating = 0
    for user in cleaned_data:
        total_rating += float(user["rating"])

    print(f"avg rating is : {total_rating/len(cleaned_data)}")

    #percentage of users with poor rating

    poor_rating = 0;
    for user in cleaned_data:
        if (float(user["rating"]) < 3.0):
            poor_rating += 1
    print(f"Percentage of users with poor rating : {poor_rating/len(cleaned_data) * 100} %")
```

```python
get_insights(cleaned_data)
```

```python
#Build recommendation feauters
def get_recommendations(data):
    recommendations = []

    for user in data:
        curr = {}
        curr["name"] = user["name"]
        curr["rating"] = user["rating"]
        if (float(user["rating"]) >= 4.0):
            curr["Brand"] = "Samsung"
        else :
            curr["Brand"] = "Other"
        recommendations.append(curr)
    return recommendations
```

```python
rec_data = get_recommendations(cleaned_data)
for user in rec_data:
    print(user)
```
