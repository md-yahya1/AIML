```python
import requests
```

```python

URL = "https://library-management-api-i6if.onrender.com/api/books"

res = requests.get(URL)
print(res.status_code)
data = res.json()
```

```python
import pandas as pd

df = pd.json_normalize(data["books"])
df = df[['_id', 'title','author','category','publishedYear']]
```
