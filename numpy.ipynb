```python
#comparing time

size = 100_000_000

py_list = list(range(size))

#python list
start = time.time()
sq_list = [x**2 for x in py_list]
end = time.time()
print(f"It satrted from {start} and ended at {end} , took {end - start} seconds")


#numpy array
np_arr = np.array(sq_list)
start = time.time()
sq_array = np_arr ** 2
end = time.time()
print(f"It satrted from {start} and ended at {end} , took {end - start} seconds")
```

```python
#comparing size

import sys

print(f"python list size = {sys.getsizeof(py_list) * len(py_list)} bytes")

print(f"python list size = {np_arr.nbytes} bytes")
```

```python
#creating array
arr = np.array([1,2,3,4,5,6])
print(arr, type(arr), arr.shape)

arr2 = np.array([1,2,3,4,5, "prime"])
print(arr2, type(arr2), arr2.dtype, arr2.shape)

arr3 = np.array([ [1,2], [4,5], [7,8]])
print(arr3, type(arr3), arr3.dtype, arr3.shape)
```

```python
#creating array from 
arr1 = np.zeros((3,3), dtype = "int64")
print(arr1, arr1.shape)

arr2 = np.ones((2,3), dtype = "int64")
print(arr2, arr2.shape)

arr3 = np.full((3,4), 100)
print(arr3, arr3.shape)

arr4 = np.eye(3, dtype = "int64")
print(arr4, arr4.shape)

arr5 = np.arange(0,11,2)
print(arr5, arr5.shape)

arr6 = np.linspace(1,100,3, dtype = "int64")
print(arr6, arr6.shape)
```

```python
#normaliztion
arr = np.array([[1,2],[3,4]])

mean = np.mean(arr)
std_dev = np.std(arr)

nrm_arr = (arr-mean)/std_dev
print(nrm_arr)
```
