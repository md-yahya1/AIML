```python
import numpy as np
```

```python
#Linear Regression

class LinearRegression:
    def __init__(self, learning_rate=0.01, n_iter=1000):
        self.bias = None
        self.weights = None
        self.lr = learning_rate
        self.n_iter = n_iter

    def fit(self, X, y): 
        # X_train, Y_train
        # (Number of Samples, Number of Feauters)
        m, n = X.shape
        
        #Step 1 (Initializze the parameters)
        self.bias = 0
        self.weights = np.zeros(n) #[0, 0, ..., 0]

        #Gradient descent
        for i in range(self.n_iter):
            #Step 2 (Calculate y_predict)
            y_pred = self.bias + np.dot(X, self.weights)
    
            #Step 3 (Find Derivatives/ Gradients)
            db = (1/m) * np.sum(y_pred - y)
            dw = (1/m) * np.dot(X.T, (y_pred-y))
    
            #Step 4 (Convergence Theorem, Update Parameters)
            self.bias -= self.lr*db
            self.weights -= self.lr*dw
            
    def predict(self, X):
        y_pred = self.bias + np.dot(X, self.weights)
        return y_pred
        
```

```python
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 6, 8, 10])

model = LinearRegression()
model.fit(X, y)

y_pred = model.predict(X)
print(y_pred)
print(model.bias)
print(model.weights)
```

```python
class LinearRegression:
    def __init__(self, learning_rate=0.01, n_iter=1000):
        self.bias = None
        self.weights = None
        self.lr = learning_rate
        self.n_iter = n_iter
        
    def fit(self, X, y): 
        # X_train, Y_train
        # (Number of Samples, Number of Feauters)
        m, n = X.shape

        X_b = np.c_[np.ones((m,1)), X]
        # theta = np.dot(np.linealg.inv(X_b.T, X_b), np.dot(X_b.T, y))
        theta = np.linalg.inv(X_b.T @ X_b) @ X_b.T @ y

        self.bias = theta[0]
        self.weights = theta[1:]
        
        

    def predict(self, X):
        y_pred = self.bias + np.dot(X, self.weights)
        return y_pred
```

```python
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 6, 8, 10])

model = LinearRegression()
model.fit(X, y)

y_pred = model.predict(X)
print(y_pred)
print(model.bias)
print(model.weights)
```
