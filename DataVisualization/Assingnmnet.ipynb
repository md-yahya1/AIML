```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.random.normal(loc=70, scale=10, size=100)
mean_score = np.mean(data)

fig, ax = plt.subplots()

ax.hist(data, bins=20, edgecolor="black")
ax.axvline(mean_score, linestyle="-", label=f"Mean = {mean_score:.2f}", color="red")

ax.set_title("Histogram of Student Scores")
ax.set_xlabel("Scores")
ax.set_ylabel("Frequency")
ax.legend()
```

```python
import seaborn as sns

data = sns.load_dataset("penguins")

sns.relplot(
    data=data,
    x="flipper_length_mm",
    y="body_mass_g",
    hue="species"
)
```

```python
tips = sns.load_dataset("tips")
sns.boxplot(
    data=tips,
    x="total_bill",
    y="day",
    hue="sex"
)

plt.title("Total Bill by Day and Gender")
plt.xlabel("Day")
plt.ylabel("Total Bill")
```

```python
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    style="smoker"
)

plt.title("Total Bill trip")
plt.xlabel("total bill")
plt.ylabel("tips")
plt.tight_layout()
```

```python

tips_pivot = tips.pivot(index="day", columns="time", values="tip",aggfunc="mean")

sns.heatmap(tips_pivot, annot=True)

plt.show()
```
