```python
import requests
from bs4 import BeautifulSoup
```

```python
base = "https://quotes.toscrape.com"
url = base
all_html = ""

while url:
    res = requests.get(url)

    all_html += res.text

    soup = BeautifulSoup(res.text, "html.parser")

    next_btn = soup.find("li", class_='next')
    
    if next_btn:
        url = base + next_btn.find("a")["href"]
    else:
        url= None

with open("data2.html", "w", encoding="utf-8") as f:
    f.write(all_html)
```

```python
data_set =[]
with open("data2.html", "r", encoding="utf-8") as f:
    html_content = f.read()

pages = html_content.split("<!DOCTYPE html>")

for page in pages:
    soup = BeautifulSoup(page, "lxml")
    
    quotes = soup.find_all("div", class_="quote")
    
    for quote in quotes:
        life_quote = quote.find("a", class_="tag", string="life")
        if life_quote:
            text = quote.find("span", class_="text")
            data = text.get_text(strip=True)
            data_set.append(data)
```

```python
import pandas as pd

df = pd.DataFrame(data_set, columns=["Life quotes"])
df
```
