```python
import requests
```

```python
URL = "https://www.scrapethissite.com/pages/simple/"
res = requests.get(URL)
```

```python
print(res.status_code)
if(res.status_code == 200):
    print(res.headers)
with open("WebScraping/ScrapedData/data1.html", "wb") as f:
    f.write(res.content)
```

```python
from bs4 import BeautifulSoup

with open("WebScraping/ScrapedData/data1.html", "r") as f:
    html_content = f.read()
soup = BeautifulSoup(html_content, "lxml")
```

```python
all_h3 = soup.find_all("h3")
data = []
for h3 in all_h3:
    name =(h3.get_text(strip=True))
    population = (h3.find_next("div").select("span.country-population")[0].get_text(strip=True))
    #  population.append(h3.find_next("div").select_one("span.country-population").get_text(strip=True))
    data.append([name, population])
   
```

```python
import pandas as pd
df = pd.DataFrame(data, columns =["Name", "population"])
```
